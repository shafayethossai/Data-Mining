```python
import numpy as np
import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules
```

```python
df = pd.read_csv('/content/drive/MyDrive/Data mining Lab/Copy of accidents.csv')
df
```

```python
import pandas as pd

df_encoded = pd.get_dummies(df, columns=df.columns)
df_encoded
```

```python
# Frequent itemset generation

frequent_itemsets = apriori(df_encoded, min_support=0.5, use_colnames=True)
print(frequent_itemsets)
```

```python
for item in frequent_itemsets.iloc[90]:
  print(item)
```

```python
# Association rule generation
num_itemsets = len(frequent_itemsets)
rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.7, num_itemsets = num_itemsets)
rules
```

```python
rules = rules.sort_values(by=['confidence', 'lift', 'support'], ascending=False)
rules
```

```python
rules[:40][['antecedents', 'consequents']]
```

```python
from mlxtend.frequent_patterns import fpgrowth

# Frequent itemset generation using FP-Growth
frequent_itemsets = fpgrowth(df_encoded, min_support=0.8, use_colnames=True)
print(frequent_itemsets)

# Association rule generation
num_itemsets = len(frequent_itemsets)
rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.7, num_itemsets = num_itemsets)
rules = rules.sort_values(by=['confidence', 'lift', 'support'], ascending=False)
rules
```
