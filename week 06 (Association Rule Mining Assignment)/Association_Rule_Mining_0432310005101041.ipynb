### 1. Setup and Data Loading

This text cell marks the beginning of the setup and data loading section of your notebook.

```python
import numpy as np
import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules, fpgrowth
from mlxtend.preprocessing import TransactionEncoder # Import TransactionEncoder
from google.colab import drive
import time

# Mount Google Drive
drive.mount('/content/drive')
```

This cell imports all the necessary libraries for the assignment:
- `numpy` and `pandas` for data manipulation.
- `apriori`, `association_rules`, and `fpgrowth` from `mlxtend.frequent_patterns` for implementing the association rule mining algorithms.
- `TransactionEncoder` from `mlxtend.preprocessing` for one-hot encoding transaction data.
- `time` for measuring execution times.
- `google.colab.drive` to mount Google Drive and access your dataset.

It then mounts your Google Drive, making your files accessible within the Colab environment.

```python
# Load the dataset
df = pd.read_csv('/content/drive/MyDrive/Data mining Lab/Association_Rule_Mining_Dataset - Association_Rule_Mining_Dataset.csv')

# Display the first few rows of the original dataframe
print("Original DataFrame:")
display(df.head())
print(f"Shape of original DataFrame: {df.shape}")
```

This cell loads your dataset from the specified CSV file located in your Google Drive into a pandas DataFrame called `df`. It then displays the first 5 rows of the loaded DataFrame and its shape, giving you an initial look at the raw data.

### 2. Data Preprocessing: One-Hot Encoding

This text cell indicates the start of the data preprocessing section, specifically focusing on converting the data into a format suitable for association rule mining using one-hot encoding.

```python
# Convert the 'Items' column into a list of lists of items (transactions)
transactions = df['Items'].apply(lambda x: x.split(', ')).tolist()

# Initialize TransactionEncoder and fit it to the transactions
te = TransactionEncoder()
te_ary = te.fit(transactions).transform(transactions)

# Convert the encoded array into a DataFrame with item names as column headers
df_encoded = pd.DataFrame(te_ary, columns=te.columns_)

# Display the first few rows of the encoded dataframe
print("One-Hot Encoded DataFrame:")
display(df_encoded.head())
print(f"Shape of encoded DataFrame: {df_encoded.shape}")
```

This cell performs the crucial preprocessing step:
1. It extracts the `Items` column from your `df` DataFrame.
2. For each row in the `Items` column, it splits the comma-separated string of items into a list of individual items, creating a list of transactions.
3. It initializes a `TransactionEncoder` object, which is then fitted to your `transactions` list.
4. The `transform` method converts your list of transactions into a one-hot encoded NumPy array, where each column represents a unique item and values are `True`/`False` indicating presence/absence.
5. Finally, this NumPy array is converted into a pandas DataFrame (`df_encoded`), using the item names as column headers. This is the required binary format for the Apriori and FP-Growth algorithms.

It then displays the first few rows and the shape of the newly created `df_encoded` DataFrame.

### 3. Apply Apriori Algorithm

This text cell signals the beginning of the section where the Apriori algorithm is applied.

```python
# Define parameters for Apriori
min_support_apriori = 0.20 # 20%
min_confidence_apriori = 0.60 # 60%

print(f"Apriori Parameters: Minimum Support = {min_support_apriori}, Minimum Confidence = {min_confidence_apriori}")

# Measure execution time for Apriori frequent itemset generation
start_time_apriori_itemsets = time.time()
frequent_itemsets_apriori = apriori(df_encoded, min_support=min_support_apriori, use_colnames=True)
end_time_apriori_itemsets = time.time()
apriori_itemsets_time = end_time_apriori_itemsets - start_time_apriori_itemsets

print(f"\nFrequent Itemsets (Apriori, Top 10):")
display(frequent_itemsets_apriori.head(10))
print(f"Number of frequent itemsets found by Apriori: {len(frequent_itemsets_apriori)}")
print(f"Time taken for Apriori frequent itemset generation: {apriori_itemsets_time:.4f} seconds")

# Generate association rules from Apriori frequent itemsets
start_time_apriori_rules = time.time()
rules_apriori = association_rules(frequent_itemsets_apriori, metric="confidence", min_threshold=min_confidence_apriori)
end_time_apriori_rules = time.time()
apriori_rules_time = end_time_apriori_rules - start_time_apriori_rules

# Sort rules by Lift, then Confidence, then Support in descending order
rules_apriori = rules_apriori.sort_values(by=['lift', 'confidence', 'support'], ascending=False)

print(f"\nAssociation Rules (Apriori, Top 10 by Lift):")
display(rules_apriori.head(10))
print(f"Number of rules found by Apriori: {len(rules_apriori)}")
print(f"Time taken for Apriori rule generation: {apriori_rules_time:.4f} seconds")
```

This cell applies the Apriori algorithm to find frequent itemsets and then generates association rules:
- It defines the `min_support_apriori` (20%) and `min_confidence_apriori` (60%) thresholds as specified in the assignment.
- It measures the time taken to find frequent itemsets using `apriori()` on `df_encoded`.
- It displays the first 10 frequent itemsets found and the total count, along with the execution time.
- It then measures the time taken to generate association rules from these frequent itemsets using `association_rules()`.
- The generated rules are sorted by `lift`, then `confidence`, and `support` in descending order to highlight the strongest rules.
- Finally, it displays the top 10 association rules found and the total count, along with the execution time for rule generation.

### 4. Apply FP-Growth Algorithm

This text cell indicates the start of the section for applying the FP-Growth algorithm.

```python
# Define parameters for FP-Growth (same as Apriori for fair comparison)
min_support_fpgrowth = 0.20 # 20%
min_confidence_fpgrowth = 0.60 # 60%

print(f"FP-Growth Parameters: Minimum Support = {min_support_fpgrowth}, Minimum Confidence = {min_confidence_fpgrowth}")

# Measure execution time for FP-Growth frequent itemset generation
start_time_fpgrowth_itemsets = time.time()
frequent_itemsets_fpgrowth = fpgrowth(df_encoded, min_support=min_support_fpgrowth, use_colnames=True)
end_time_fpgrowth_itemsets = time.time()
fpgrowth_itemsets_time = end_time_fpgrowth_itemsets - start_time_fpgrowth_itemsets

print(f"\nFrequent Itemsets (FP-Growth, Top 10):")
display(frequent_itemsets_fpgrowth.head(10))
print(f"Number of frequent itemsets found by FP-Growth: {len(frequent_itemsets_fpgrowth)}")
print(f"Time taken for FP-Growth frequent itemset generation: {fpgrowth_itemsets_time:.4f} seconds")

# Generate association rules from FP-Growth frequent itemsets
start_time_fpgrowth_rules = time.time()
rules_fpgrowth = association_rules(frequent_itemsets_fpgrowth, metric="confidence", min_threshold=min_confidence_fpgrowth)
end_time_fpgrowth_rules = time.time()
fpgrowth_rules_time = end_time_fpgrowth_rules - start_time_fpgrowth_rules

# Sort rules by Lift, then Confidence, then Support in descending order
rules_fpgrowth = rules_fpgrowth.sort_values(by=['lift', 'confidence', 'support'], ascending=False)

print(f"\nAssociation Rules (FP-Growth, Top 10 by Lift):")
display(rules_fpgrowth.head(10))
print(f"Number of rules found by FP-Growth: {len(rules_fpgrowth)}")
print(f"Time taken for FP-Growth rule generation: {fpgrowth_rules_time:.4f} seconds")
```

This cell applies the FP-Growth algorithm, similar to the Apriori cell, to find frequent itemsets and generate association rules:
- It uses the same `min_support_fpgrowth` (20%) and `min_confidence_fpgrowth` (60%) thresholds for a fair comparison.
- It measures the time taken to find frequent itemsets using `fpgrowth()` on `df_encoded`.
- It displays the first 10 frequent itemsets found and the total count, along with the execution time.
- It then measures the time taken to generate association rules from these frequent itemsets using `association_rules()`.
- The generated rules are also sorted by `lift`, then `confidence`, and `support` in descending order.
- Finally, it displays the top 10 association rules found and the total count, along with the execution time for rule generation.

### 5. Comparison of Apriori and FP-Growth

This text cell indicates the start of the section for comparing the two algorithms.

```python
print("\n--- Comparison --- ")
print(f"Number of Apriori Frequent Itemsets: {len(frequent_itemsets_apriori)}")
print(f"Number of FP-Growth Frequent Itemsets: {len(frequent_itemsets_fpgrowth)}")
print(f"Number of Apriori Rules: {len(rules_apriori)}")
print(f"Number of FP-Growth Rules: {len(rules_fpgrowth)}")

print(f"\nApriori Total Time (Itemsets + Rules): {apriori_itemsets_time + apriori_rules_time:.4f} seconds")
print(f"FP-Growth Total Time (Itemsets + Rules): {fpgrowth_itemsets_time + fpgrowth_rules_time:.4f} seconds")

# Identify the strongest rule (based on lift) and explain its business meaning.
print("\n--- Strongest Rule and Business Meaning (Apriori) ---")
if not rules_apriori.empty:
    strongest_rule_apriori = rules_apriori.iloc[0]
    print("Strongest Apriori Rule (by Lift):")
    print(strongest_rule_apriori)

    antecedents_str = ', '.join(list(strongest_rule_apriori['antecedents']))
    consequents_str = ', '.join(list(strongest_rule_apriori['consequents']))

    print("\nBusiness Meaning:")
    print(f"If a transaction contains '{antecedents_str}', then it is highly likely to also contain '{consequents_str}'.")
    print(f"This rule has a support of {strongest_rule_apriori['support']:.2%}, meaning it appears in {strongest_rule_apriori['support']:.2%} of all transactions.")
    print(f"The confidence is {strongest_rule_apriori['confidence']:.2%}, indicating that {strongest_rule_apriori['confidence']:.2%} of transactions with '{antecedents_str}' also contain '{consequents_str}'.")
    print(f"The lift value of {strongest_rule_apriori['lift']:.2f} suggests that the presence of '{antecedents_str}' increases the likelihood of '{consequents_str}' by a factor of {strongest_rule_apriori['lift']:.2f} compared to its baseline probability.")
else:
    print("No Apriori rules found to analyze.")
```

This cell provides a comprehensive comparison and analysis:
- It prints the number of frequent itemsets and rules found by both Apriori and FP-Growth.
- It summarizes the total execution time for each algorithm (itemset generation + rule generation).
- It identifies the strongest association rule based on the 'lift' metric from the Apriori results. Lift indicates how much more likely the consequent is given the antecedent than without it. A lift > 1 suggests a positive correlation.
- It then provides a detailed business interpretation of this strongest rule, explaining its support, confidence, and lift values in practical terms.

### 6. Conclusion

This text cell introduces the final conclusion section of your analysis.

```python
print("\n--- Conclusion ---")

# Compare based on execution time and number of itemsets/rules
if (apriori_itemsets_time + apriori_rules_time) < (fpgrowth_itemsets_time + fpgrowth_rules_time):
    efficiency_conclusion = "Apriori appears to be slightly more efficient for this dataset in terms of execution time."
elif (apriori_itemsets_time + apriori_rules_time) > (fpgrowth_itemsets_time + fpgrowth_rules_time):
    efficiency_conclusion = "FP-Growth appears to be slightly more efficient for this dataset in terms of execution time."
else:
    efficiency_conclusion = "Both algorithms show similar efficiency for this dataset in terms of execution time."

print(f"\nComputational Efficiency: {efficiency_conclusion}")

if len(frequent_itemsets_apriori) == len(frequent_itemsets_fpgrowth) and len(rules_apriori) == len(rules_fpgrowth):
    print("Both algorithms found the same number of frequent itemsets and association rules, as expected given they discover all frequent itemsets that satisfy the minimum support.")
else:
    print("There was a difference in the number of frequent itemsets or rules found, which might indicate a slight discrepancy in handling edge cases or internal optimizations.")

print("\nOverall, both algorithms successfully identified frequent patterns and generated actionable association rules from the supermarket transaction data. The choice between Apriori and FP-Growth often depends on the dataset's characteristics, especially its density and size. For larger and denser datasets, FP-Growth typically outperforms Apriori due to its tree-based structure which avoids repeated candidate generation.")
```

This cell concludes the analysis by summarizing the computational efficiency of both algorithms for this specific dataset:
